```python
from sklearn.datasets import load_iris

# Load the iris dataset
iris = load_iris()

# Display the keys of the dataset to understand its structure
print(iris.keys())
```

```python
x = iris.data
y = iris.target

print("Shape of x:", x.shape)
print("Shape of y:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

# Perform a train-test split with a test size of 20% and a random state for reproducibility
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

print("Shape of X_train:", X_train.shape)
print("Shape of X_test:", X_test.shape)
print("Shape of y_train:", y_train.shape)
print("Shape of y_test:", y_test.shape)
```

```python
from sklearn.naive_bayes import GaussianNB

# Initialize the Gaussian Naive Bayes model
gnb = GaussianNB()

# Train the model using the training data
gnb.fit(X_train, y_train)

print("Gaussian Naive Bayes model trained successfully.")
```

```python
# Make predictions on the test data
y_pred = gnb.predict(X_test)

print("Predictions for X_test:", y_pred)
```

```python
from sklearn.metrics import accuracy_score

# Calculate the accuracy score
accuracy = accuracy_score(y_test, y_pred)

print(f"Accuracy Score: {accuracy*100}%")
```

```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("All requested libraries imported successfully.")
```

```python
# Load the wine dataset
wine = load_wine()

# Display the keys of the dataset to understand its structure
print(wine.keys())
```

```python
import pandas as pd

# Convert the wine dataset to a pandas DataFrame for easier manipulation
wine_df = pd.DataFrame(data=wine.data, columns=wine.feature_names)
wine_df['target'] = wine.target

print("Wine dataset converted to DataFrame.")
display(wine_df.head())
```

```python
# Assign features (X) and target (y) for the wine dataset
X_wine = wine.data
y_wine = wine.target

print("Shape of X_wine:", X_wine.shape)
print("Shape of y_wine:", y_wine.shape)
```

```python
from sklearn.model_selection import train_test_split

# Perform a train-test split with a test size of 30% and a random state for reproducibility
X_train_wine, X_test_wine, y_train_wine, y_test_wine = train_test_split(X_wine, y_wine, test_size=0.3, random_state=42)

print("Shape of X_train_wine:", X_train_wine.shape)
print("Shape of X_test_wine:", X_test_wine.shape)
print("Shape of y_train_wine:", y_train_wine.shape)
print("Shape of y_test_wine:", y_test_wine.shape)
```

```python
from sklearn.naive_bayes import GaussianNB

# Initialize the Gaussian Naive Bayes model for the wine dataset
gnb_wine = GaussianNB()

# Train the model using the training data of the wine dataset
gnb_wine.fit(X_train_wine, y_train_wine)

print("Gaussian Naive Bayes model trained successfully for the wine dataset.")
```

```python
# Make predictions on the test data for the wine dataset
y_pred_wine = gnb_wine.predict(X_test_wine)

print("Predictions for X_test_wine:", y_pred_wine)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix

# Calculate the accuracy score for the wine dataset
accuracy_wine = accuracy_score(y_test_wine, y_pred_wine)
print(f"Accuracy Score for Wine Dataset: {accuracy_wine*100:.2f}%")

# Calculate the confusion matrix for the wine dataset
cm_wine = confusion_matrix(y_test_wine, y_pred_wine)
print("\nConfusion Matrix for Wine Dataset:\n", cm_wine)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Plot the confusion matrix
plt.figure(figsize=(8, 6))
sns.heatmap(cm_wine, annot=True, fmt="d", cmap="Blues", cbar=False,
            xticklabels=wine.target_names, yticklabels=wine.target_names)
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix for Wine Dataset (Gaussian Naive Bayes)')
plt.savefig('confusion_matrix_wine.png')
plt.show()
```
